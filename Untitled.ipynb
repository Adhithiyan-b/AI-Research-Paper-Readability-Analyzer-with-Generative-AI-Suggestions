```python
# 3. Build Custom Validation Split and Organize Directory Structure
import shutil
import glob
import random
from pathlib import Path

# Define paths
base_dir = Path('./chest_xray_raw/chest_xray')
train_dir = base_dir / 'train'
val_dir = base_dir / 'val' # original val set (very small)
test_dir = base_dir / 'test'

# Create a new custom validation split directory
custom_val_dir = Path('./dataset/val')
custom_train_dir = Path('./dataset/train')
custom_test_dir = Path('./dataset/test')

for d in [custom_train_dir, custom_val_dir, custom_test_dir]:
    for category in ['NORMAL', 'PNEUMONIA']:
        (d / category).mkdir(parents=True, exist_ok=True)

# Function to copy files
def copy_files(files, dest_folder):
    for f in files:
        shutil.copy(f, dest_folder / Path(f).name)

# Split training files into 85% train and 15% validation to create a robust validation set
for category in ['NORMAL', 'PNEUMONIA']:
    train_files = glob.glob(str(train_dir / category / '*'))
    random.seed(42)
    random.shuffle(train_files)

    split_idx = int(len(train_files) * 0.85)
    train_split = train_files[:split_idx]
    val_split = train_files[split_idx:]

    copy_files(train_split, custom_train_dir / category)
    copy_files(val_split, custom_val_dir / category)

    # Copy test files
    test_files = glob.glob(str(test_dir / category / '*'))
    copy_files(test_files, custom_test_dir / category)

# Print dataset distribution summary
print("Dataset Splitting Summary:")
for d_name, d_path in [('Train', custom_train_dir), ('Val (Custom)', custom_val_dir), ('Test', custom_test_dir)]:
    normal_cnt = len(list((d_path / 'NORMAL').glob('*')))
    pneumonia_cnt = len(list((d_path / 'PNEUMONIA').glob('*')))
    total = normal_cnt + pneumonia_cnt
    print(f"{d_name} -> Normal: {normal_cnt}, Pneumonia: {pneumonia_cnt}, Total: {total}")
```

```python
import torch
import torch.nn as nn
import torch.optim as optim
from torchvision import datasets, transforms, models
from torch.utils.data import DataLoader
import matplotlib.pyplot as plt
import numpy as np
from sklearn.metrics import classification_report, confusion_matrix, recall_score, accuracy_score

# Set device
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f'Using device: {device}')
```

```python
# 4. Image Preprocessing & Augmentation
# Since Pneumonia class has more images than Normal, we add slight augmentations to train data
train_transform = transforms.Compose([
    transforms.Resize((150, 150)),
    transforms.RandomRotation(10),
    transforms.RandomHorizontalFlip(),
    transforms.ColorJitter(brightness=0.1, contrast=0.1),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225])
])

val_test_transform = transforms.Compose([
    transforms.Resize((150, 150)),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225])
])

# Datasets
train_dataset = datasets.ImageFolder('./dataset/train', transform=train_transform)
val_dataset = datasets.ImageFolder('./dataset/val', transform=val_test_transform)
test_dataset = datasets.ImageFolder('./dataset/test', transform=val_test_transform)

# Data Loaders
batch_size = 32
train_loader = DataLoader(train_dataset, batch_size=batch_size, shuffle=True, num_workers=2)
val_loader = DataLoader(val_dataset, batch_size=batch_size, shuffle=False, num_workers=2)
test_loader = DataLoader(test_dataset, batch_size=batch_size, shuffle=False, num_workers=2)

print(f"Classes mapping: {train_dataset.class_to_idx}")
```

```python
# 5. Model Architecture (Transfer Learning using ResNet18)
model = models.resnet18(pretrained=True)

# Freeze feature layers
for param in model.parameters():
    param.requires_grad = False

# Replace final fully connected layer to classify 2 classes (Normal, Pneumonia)
num_ftrs = model.fc.in_features
model.fc = nn.Sequential(
    nn.Linear(num_ftrs, 128),
    nn.ReLU(),
    nn.Dropout(0.3),
    nn.Linear(128, 2)
)

model = model.to(device)

# Loss function and Optimizer
criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.fc.parameters(), lr=1e-3)
```

```python
# 6. Training and Validation Loop (10 Epochs)
epochs = 10
train_losses, val_losses = [], []
train_accs, val_accs = [], []

for epoch in range(epochs):
    model.train()
    running_loss, correct, total = 0.0, 0, 0

    for images, labels in train_loader:
        images, labels = images.to(device), labels.to(device)

        optimizer.zero_grad()
        outputs = model(images)
        loss = criterion(outputs, labels)
        loss.backward()
        optimizer.step()

        running_loss += loss.item() * images.size(0)
        _, predicted = outputs.max(1)
        total += labels.size(0)
        correct += predicted.eq(labels).sum().item()

    epoch_train_loss = running_loss / len(train_dataset)
    epoch_train_acc = correct / total

    # Validation
    model.eval()
    running_val_loss, val_correct, val_total = 0.0, 0, 0
    with torch.no_grad():
        for images, labels in val_loader:
            images, labels = images.to(device), labels.to(device)
            outputs = model(images)
            loss = criterion(outputs, labels)

            running_val_loss += loss.item() * images.size(0)
            _, predicted = outputs.max(1)
            val_total += labels.size(0)
            val_correct += predicted.eq(labels).sum().item()

    epoch_val_loss = running_val_loss / len(val_dataset)
    epoch_val_acc = val_correct / val_total

    train_losses.append(epoch_train_loss)
    val_losses.append(epoch_val_loss)
    train_accs.append(epoch_train_acc)
    val_accs.append(epoch_val_acc)

    print(f"Epoch {epoch+1}/{epochs} - Train Loss: {epoch_train_loss:.4f}, Train Acc: {epoch_train_acc:.4f} | Val Loss: {epoch_val_loss:.4f}, Val Acc: {epoch_val_acc:.4f}")
```

```python
# 7. Plot Loss and Accuracy Metrics
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
plt.plot(train_losses, label='Train Loss')
plt.plot(val_losses, label='Val Loss')
plt.title('Loss Progression')
plt.xlabel('Epoch')
plt.legend()

plt.subplot(1, 2, 2)
plt.plot(train_accs, label='Train Acc')
plt.plot(val_accs, label='Val Acc')
plt.title('Accuracy Progression')
plt.xlabel('Epoch')
plt.legend()
plt.show()
```

```python
# 8. Test Evaluation: Report Accuracy & Sensitivity (Recall for Pneumonia)
model.eval()
all_preds = []
all_labels = []

with torch.no_grad():
    for images, labels in test_loader:
        images = images.to(device)
        outputs = model(images)
        _, predicted = outputs.max(1)
        all_preds.extend(predicted.cpu().numpy())
        all_labels.extend(labels.numpy())

# Map indices back to class names
class_names = ['NORMAL', 'PNEUMONIA']
accuracy = accuracy_score(all_labels, all_preds)

# Sensitivity (Recall) for Pneumonia (Class index 1)
sensitivity = recall_score(all_labels, all_preds, pos_label=1)

print("=== FINAL TEST RESULTS ===")
print(f"Accuracy: {accuracy:.4f}")
print(f"Sensitivity (Recall for Pneumonia): {sensitivity:.4f}")
print("\nConfusion Matrix:")
print(confusion_matrix(all_labels, all_preds))
print("\nClassification Report:")
print(classification_report(all_labels, all_preds, target_names=class_names))
```

```python
import seaborn as sns
from sklearn.metrics import roc_curve, auc, precision_recall_fscore_support

# 1. Calculate Evaluation Metrics on the Test Set
tn, fp, fn, tp = confusion_matrix(all_labels, all_preds).ravel()

specificity = tn / (tn + fp)
precision, recall, f1, _ = precision_recall_fscore_support(all_labels, all_preds, average='binary', pos_label=1)

print("=== EXTENDED TEST METRICS ===")
print(f"Accuracy:    {accuracy:.4f}")
print(f"Sensitivity (Recall): {sensitivity:.4f}")
print(f"Specificity: {specificity:.4f}")
print(f"Precision:   {precision:.4f}")
print(f"F1 Score:    {f1:.4f}")

# Plot Confusion Matrix Heatmap
plt.figure(figsize=(6, 5))
sns.heatmap(confusion_matrix(all_labels, all_preds), annot=True, fmt='d', cmap='Blues',
            xticklabels=class_names, yticklabels=class_names)
plt.title('Confusion Matrix Heatmap')
plt.ylabel('Actual')
plt.xlabel('Predicted')
plt.show()
```

```python
# 2. Calculate and Plot ROC Curve with AUC
model.eval()
all_probs = []

with torch.no_grad():
    for images, _ in test_loader:
        images = images.to(device)
        outputs = model(images)
        probs = torch.softmax(outputs, dim=1)[:, 1]
        all_probs.extend(probs.cpu().numpy())

fpr, tpr, thresholds = roc_curve(all_labels, all_probs)
roc_auc = auc(fpr, tpr)

plt.figure(figsize=(7, 6))
plt.plot(fpr, tpr, color='darkorange', lw=2, label=f'ROC curve (AUC = {roc_auc:.4f})')
plt.plot([0, 1], [0, 1], color='navy', lw=2, linestyle='--')
plt.xlim([0.0, 1.0])
plt.ylim([0.0, 1.05])
plt.xlabel('False Positive Rate (1 - Specificity)')
plt.ylabel('True Positive Rate (Sensitivity)')
plt.title('Receiver Operating Characteristic (ROC) Curve')
plt.legend(loc="lower right")
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

```python
# 10. Identify and Display Misclassified Images
import torch.nn.functional as F

model.eval()
misclassified_fp = [] # NORMAL predicted as PNEUMONIA
misclassified_fn = [] # PNEUMONIA predicted as NORMAL

with torch.no_grad():
    for images, labels in test_loader:
        images_dev = images.to(device)
        outputs = model(images_dev)
        probs = F.softmax(outputs, dim=1)
        _, predicted = outputs.max(1)

        for i in range(images.size(0)):
            true_lbl = labels[i].item()
            pred_lbl = predicted[i].item()
            confidence = probs[i][pred_lbl].item()

            # Unnormalize image for plotting
            img = images[i].cpu().numpy().transpose((1, 2, 0))
            mean = np.array([0.485, 0.456, 0.406])
            std = np.array([0.229, 0.224, 0.225])
            img = std * img + mean
            img = np.clip(img, 0, 1)

            if true_lbl == 0 and pred_lbl == 1:
                if len(misclassified_fp) < 4:
                    misclassified_fp.append((img, true_lbl, pred_lbl, confidence))
            elif true_lbl == 1 and pred_lbl == 0:
                if len(misclassified_fn) < 4:
                    misclassified_fn.append((img, true_lbl, pred_lbl, confidence))

        if len(misclassified_fp) >= 4 and len(misclassified_fn) >= 4:
            break

# Plot the misclassified images
fig, axes = plt.subplots(2, 4, figsize=(16, 8))

# Row 1: False Positives
for idx, (img, true_lbl, pred_lbl, conf) in enumerate(misclassified_fp):
    ax = axes[0, idx]
    ax.imshow(img)
    ax.set_title(f"True: {class_names[true_lbl]}\nPred: {class_names[pred_lbl]}\nConf: {conf:.2%}", color='red')
    ax.axis('off')

# Row 2: False Negatives
for idx, (img, true_lbl, pred_lbl, conf) in enumerate(misclassified_fn):
    ax = axes[1, idx]
    ax.imshow(img)
    ax.set_title(f"True: {class_names[true_lbl]}\nPred: {class_names[pred_lbl]}\nConf: {conf:.2%}", color='red')
    ax.axis('off')

plt.tight_layout()
plt.show()
```

### 11. Visual Observations of Misclassified X-rays

Based on the visual analysis of the misclassified chest X-ray samples, we observe the following common characteristics:

1. **Rotation and Positioning Variations**: Several misclassified images exhibit significant rotation or slight lateral tilt of the child's torso. This structural shift alters the traditional bounding box reference of the lungs, potentially confusing the convolutional feature maps.
2. **Low Contrast and Under-exposure**: Some images suffer from poor contrast or under-exposure (highly dark regions), making it visually challenging to distinguish the delicate vascular markings of a normal lung from the consolidation/infiltrates characteristic of pneumonia.
3. **Thymus Shadow & Rib Congestion**: In young pediatric patients, a prominent thymus shadow or closely spaced ribs can occasionally overlay the lung field, mimicking the increased opacity associated with lobar or bronchopneumonia.
